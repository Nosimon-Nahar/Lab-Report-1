```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
image = cv2.imread('/content/capybara.jpg')
image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)
```

```python
# image dimension
h, w = image.shape[:2]
```

```python
# Step 1: Scaling
```

```python
sx= 3
sy= 1.5
```

```python
scaled = cv2.resize(image, None, fx=sx, fy=sy)
```

```python
# Step 2: Rotation
```

```python
center= (w//2, h//2)
```

```python
M_rotation = cv2.getRotationMatrix2D(center, 40, 1.0)
```

```python
rotated= cv2.warpAffine(image, M_rotation, (w, h))
```

```python
# step 3: Translation
```

```python
# translation value
tx =50
ty =30
```

```python
# translation matrix
M_translation = np.float32([
    [1, 0, tx],
    [0, 1, ty]
])
```

```python
translated= cv2.warpAffine(image, M_translation, (w,h))
```

```python
#step 4: reflected
```

```python
reflected = cv2.flip(image, 0)
```

```python
#display
plt.figure(figsize=(15,5))
```

```python
plt.subplot(2,3,1)
plt.imshow(image)
plt.title('Original Image')
plt.axis('off')

plt.subplot(2,3,2)
plt.imshow(translated)
plt.title('Translated')
plt.axis('off')

plt.subplot(2,3,3)
plt.imshow(scaled)
plt.title('Scaled')
plt.axis('off')

plt.subplot(2,3,4)
plt.imshow(rotated)
plt.title('Rotated')
plt.axis('off')

plt.subplot(2,3,5)
plt.imshow(reflected)
plt.title('Reflected')
plt.axis('off')

plt.show()
```
