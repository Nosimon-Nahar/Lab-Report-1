```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
image = cv2.imread('/content/capybara.jpg')
image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)
```

```python
# image dimension
h, w = image.shape[:2] # (height, width), channel
```

```python
# translation value
tx =100
ty =50
```

```python
# translation matrix
M = np.float32([
    [1, 0, tx],
    [0, 1, ty]
])
```

```python
translatedImage= cv2.warpAffine(image, M, (w,h))
```

```python
plt.figure(figsize=(20,15))
```

```python
plt.subplot(1,2,1)
plt.imshow(image)
plt.title('Original')
plt.axis('off')


plt.subplot(1,2,2)
plt.imshow(translatedImage)
plt.title('Translated')
plt.axis('off')

plt.show()
```
